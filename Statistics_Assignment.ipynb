## Question 1
**What is hypothesis testing in statistics?**

Hypothesis testing is a statistical method used to make decisions or inferences about population parameters based on sample data. It involves formulating a null hypothesis and an alternative hypothesis, selecting a significance level, computing a test statistic, and deciding whether to reject or fail to reject the null hypothesis.

## Question 2
**What is the null hypothesis, and how does it differ from the alternative hypothesis?**

The null hypothesis (H0) states that there is no effect, no difference, or no relationship in the population. The alternative hypothesis (H1 or Ha) states that there is an effect, a difference, or a relationship. The goal of hypothesis testing is to test evidence against the null hypothesis.

## Question 3
**Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.**

The significance level (α) is the probability of rejecting the null hypothesis when it is actually true. Common values are 0.05 or 0.01. If the p-value is less than α, we reject the null hypothesis; otherwise, we fail to reject it.

## Question 4
**What are Type I and Type II errors? Give examples of each.**

Type I error: Rejecting a true null hypothesis. Example: Concluding a drug works when it actually does not.

Type II error: Failing to reject a false null hypothesis. Example: Concluding a drug does not work when it actually does.

## Question 5
**What is the difference between a Z-test and a T-test?**

A Z-test is used when the population variance is known and the sample size is large (n ≥ 30). A T-test is used when the population variance is unknown and the sample size is small. The T-test uses the t-distribution instead of the normal distribution.

## Question 6
**Generate a binomial distribution and plot its histogram.**

```
import numpy as np
import matplotlib.pyplot as plt

n = 10
p = 0.5
data = np.random.binomial(n, p, size=1000)

plt.hist(data, bins=range(n+2), edgecolor='black')
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.show()
```

## Question 7
**Hypothesis testing using Z-statistics.**

```
import numpy as np
from scipy import stats

sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

mu0 = 50
sigma = np.std(sample_data, ddof=1)
x_bar = np.mean(sample_data)

z = (x_bar - mu0) / (sigma / np.sqrt(len(sample_data)))
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

z, p_value
```

Interpretation: If the p-value is less than 0.05, reject the null hypothesis; otherwise, fail to reject it.

## Question 8
**Normal distribution simulation and 95% confidence interval.**

```
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

data = np.random.normal(loc=50, scale=5, size=1000)

mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

ci = stats.norm.interval(0.95, loc=mean, scale=std/np.sqrt(n))

plt.hist(data, bins=30, edgecolor='black')
plt.title("Normal Distribution Simulation")
plt.show()

mean, ci
```

## Question 9
**Z-score calculation and visualization.**

```
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 5, 500)

mean = np.mean(data)
std = np.std(data)

z_scores = (data - mean) / std

plt.hist(z_scores, bins=30, edgecolor='black')
plt.title("Histogram of Z-scores")
plt.show()

z_scores[:10]
```

Explanation: Z-scores represent how many standard deviations a data point is from the mean. Positive values indicate above the mean, negative values indicate below the mean.
